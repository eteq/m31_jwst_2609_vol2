The goal here is to do a *single* forward model that could be compared to a single MSA exposure.  The intent here is to work out all the steps, get rough timing, and use that to determine which need further optimization.  All relevant notebooks start with `single_forward_model`.

This comes after the PSF grid was generated in [single_forward_model3_psfgrid.ipynb](single_forward_model3_psfgrid.ipynb)

## Instead try a PSF photometry-like approach

Use a single PSF based on the center for now to simplify things

```python
t1 = time.time()

model_slit = model_slit_templ.copy()

psfi = psf_interpolator(psf_wl_grid[psf_wl_grid.shape[0]//2])
psfi = psf.ImagePSF(psfi*psf_oversampling*psf_oversampling,
             oversampling=psf_oversampling,
             origin=psf_center_os)

t2 = time.time()

dwl = 0.1*u.angstrom
x1, y1 = swcs.world_to_pixel(scoord, SpectralCoord(model_wave))
x2, y2 = swcs.world_to_pixel(scoord, SpectralCoord(model_wave+dwl))
det_disp = np.hypot(x2-x1, y2-y1)*u.pixel/dwl

model_flux_fnu_dispcorr = model_flux_fnu * np.gradient(model_wave)/u.pixel * det_disp

t3 = time.time()
for fnuish, wl, xc, yc in tqdm(list(zip(model_flux_fnu_dispcorr, model_wave, center_x, center_y))):
    if np.isfinite(xc) and np.isfinite(yc):
        psfi.x_0 = xc
        psfi.y_0 = yc
        psfi.flux = fnuish.value
        psfi.render(model_slit)

t4 = time.time()


plt.figure(figsize=(15, 6))
sc = viz.imshow_norm(model_slit,
                stretch=viz.AsinhStretch(), interval=viz.AsymmetricPercentileInterval(50, 99.99), 
                aspect='auto', origin='lower', interpolation='nearest')[0]
plt.plot(center_x, center_y, color='k', alpha=0.5)
plt.plot(center_x, center_y + 6*model_flux_fnu/np.max(model_flux_fnu), color='r', alpha=0.5)
plt.colorbar(sc)

t2-t1, t3-t2, t4-t3
```

## Drizzleish/pixel-overlap approach

The below is working out pieces of the broadcasting

```python
xpxi=100
ypxi = 20

xs = psf_source_offset_x_px + 100
ys = psf_source_offset_y_px + 20

fx = np.clip(psf_oversampling*-np.abs(xs-xpxi) + (1+psf_oversampling)/2, 0, 1)
fy = np.clip(psf_oversampling*-np.abs(ys-ypxi) + (1+psf_oversampling)/2, 0, 1)

plt.imshow(fx*fy, interpolation='nearest')
plt.colorbar()
plt.xlim(50, 100)
plt.ylim(75, 125)
```

```python
ypsf, xpsf = np.indices(psf_source_offset_x_px.shape)

ydat, xdat = np.indices((int(np.ceil(np.ptp(ys))), int(np.ceil(np.ptp(xs)))))
ydat += int(np.round(np.min(ys)))
xdat += int(np.round(np.min(xs)))

dx = np.subtract.outer(xs, xdat)
dy = np.subtract.outer(ys, ydat)

fx = np.clip(psf_oversampling*-np.abs(dx) + (1+psf_oversampling)/2, 0, 1)
fy = np.clip(psf_oversampling*-np.abs(dy) + (1+psf_oversampling)/2, 0, 1)
f = fx*fy
```

```python

dwl = 0.1*u.angstrom
x1, y1 = swcs.world_to_pixel(scoord, SpectralCoord(model_wave))
x2, y2 = swcs.world_to_pixel(scoord, SpectralCoord(model_wave+dwl))
det_disp = np.hypot(x2-x1, y2-y1)*u.pixel/dwl

model_flux_fnu_dispcorr = model_flux_fnu * np.gradient(model_wave)/u.pixel * det_disp

model_slit = model_slit_templ.copy()

ypsf, xpsf = np.indices(psf_source_offset_x_px.shape)
for i, (fnuish, wl, xc, yc) in tqdm(list(enumerate(zip(model_flux_fnu_dispcorr, model_wave, center_x, center_y)))):
    psfi = psf_interpolator(wl.to(psf_wl_grid.unit))
    
    xs = psf_source_offset_x_px + xc
    ys = psf_source_offset_y_px + yc

    if not (np.any(np.isfinite(xs)) and np.any(np.isfinite(ys))):
        continue    

    ydat, xdat = np.indices((int(np.floor(np.ptp(ys))), int(np.floor(np.ptp(xs)))))
    ydat += int(np.round(np.min(ys)))
    xdat += int(np.round(np.min(xs)))

    dx = np.subtract.outer(xs, xdat)
    dy = np.subtract.outer(ys, ydat)



    fx = np.clip(psf_oversampling*-np.abs(dx) + (1+psf_oversampling)/2, 0, 1)
    fy = np.clip(psf_oversampling*-np.abs(dy) + (1+psf_oversampling)/2, 0, 1)
    f = fx*fy * fnuish.value

    pp = psfi[:,:, np.newaxis, np.newaxis] * f
    dattoadd = np.sum(pp, axis=(0, 1))
    

    xyinds = np.array((ydat.ravel(), xdat.ravel()))

    msk = (xdat >= 0) & (ydat >= 0) & (xdat < model_slit.shape[1]) & (ydat < model_slit.shape[0])

    np.add.at(model_slit, (ydat[msk], xdat[msk]), dattoadd[msk])
```
