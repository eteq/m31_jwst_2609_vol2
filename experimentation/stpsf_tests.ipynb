# offset versions of the psf

```python
nrs.image_mask = 'Single MSA open shutter'
```

```python
nrs.options.update({'source_offset_x': 0., 'source_offset_y': 0.})
plt.figure(figsize=(15, 12))
psf1 = nrs.calc_psf(monochromatic=1.1e-6, oversample=8, display=True)
```

```python
nrs.options.update({'source_offset_x': 0.05, 'source_offset_y': 0.})
plt.figure(figsize=(15, 12))
psf2 = nrs.calc_psf(monochromatic=1.1e-6, oversample=8, display=True)
```

```python
nrs.options.update({'source_offset_x': 0., 'source_offset_y': 0.2})
plt.figure(figsize=(15, 12))
psf3 = nrs.calc_psf(monochromatic=1.1e-6, oversample=8, display=True)
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
for ax, psf in zip(axs, [psf1, psf2, psf3]):
    viz.imshow_norm(psf['OVERDIST'].data, stretch=viz.LogStretch(1e6), interval=viz.PercentileInterval(99.9), ax=ax, interpolation='nearest', origin='lower')
    ax.scatter([psf['OVERDIST'].data.shape[0]/2-0.5], [psf['OVERDIST'].data.shape[1]/2-0.5],c='r',s=2)
```

Some important take-aways:

* `source_offset_x` and `source_offset_y` are indeed in arcsec
* the center of the light *does* move based on source offset so that has to be accounted for by re-centering or using a different ra/dec track

# PSF interpolation tests

```python
nrs = stpsf.NIRSpec()
nrs.detector = n6791dms[0].meta.instrument.detector
nrs.filter = n6791dms[0].meta.instrument.filter
nrs.disperser = n6791dms[0].meta.instrument.grating
nrs.image_mask = 'Single MSA open shutter'
nrs.options['source_offset_x'] = nrs.options['source_offset_y'] = 0.0
```

```python
specsinimg = swcs.pixel_to_world(*np.indices(s.data.shape)[::-1])[-1]
speclims = np.min(specsinimg[np.isfinite(specsinimg)]), np.max(specsinimg[np.isfinite(specsinimg)])

wls = np.linspace(speclims[0], speclims[1], 9)
psfs = [nrs.calc_psf(monochromatic=wl.to(u.m).value, oversample=4) for wl in tqdm(wls)]
```

```python
fig, axs = plt.subplots(3, 3, figsize=(15, 15))

for wl, psf, ax in zip(wls, psfs, axs.ravel()):
    viz.imshow_norm(psf['OVERDIST'].data, stretch=viz.LogStretch(1e6), interval=viz.PercentileInterval(99.9), ax=ax, interpolation='nearest', origin='lower')
    ax.set_title(f'{wl:.2f}')
```

```python
fig, axs = plt.subplots(3, 3, figsize=(15, 15))

first = None
for wl, psf, ax in zip(wls, psfs, axs.ravel()):
    if first is None:
        first = psf['OVERDIST'].data
        toshow = first/np.sum(first)
        ax.set_title(f'{wl:.2f}')
        firstlow, firsthigh = np.quantile(toshow[np.isfinite(toshow)], [0.01, 0.99])
        firstlims = np.max(np.abs([firstlow, firsthigh]))
        ax.imshow(toshow, vmin=-firstlims, vmax=firstlims, interpolation='nearest', origin='lower')
    else:
        thisone = psf['OVERDIST'].data
        toshow = thisone/np.sum(thisone) - first/np.sum(first)
        ax.set_title(f'{wl:.2f} diff from first')
        ax.imshow(toshow, vmin=-firstlims, vmax=firstlims, interpolation='nearest', origin='lower')
```

```python
fig, axs = plt.subplots(3, 3, figsize=(15, 15))

first = None
for wl, psf, ax in zip(wls, psfs, axs.ravel()):
    if first is None:
        first = psf['OVERDIST'].data
        toshow = first
        ax.set_title(f'{wl:.2f}')
    else:
        toshow = psf['OVERDIST'].data - first
        ax.set_title(f'{wl:.2f} diff from first')
    viz.imshow_norm(toshow, stretch=viz.LogStretch(1e6), interval=viz.PercentileInterval(99.5), ax=ax, interpolation='nearest', origin='lower')
```

```python
for wl, psf in zip(wls, psfs):
    y, x = np.indices(psf['OVERDIST'].data.shape)
    dx = x - psf['OVERDIST'].data.shape[1]/2 + 0.5
    dy = y - psf['OVERDIST'].data.shape[0]/2 + 0.5
    r = np.sqrt(dx**2 + dy**2)

    sorti = np.argsort(r.ravel())
    rsort = r.ravel()[sorti]
    psfsorti = psf['OVERDIST'].data.ravel()[sorti]
    plt.scatter(rsort, psfsorti, label=f'{wl:.2f}', s=1, alpha=.3)

    plt.legend(loc=0)

    plt.ylabel('EE')
    plt.xlabel('Radius (pixels)')
    plt.xlim(0,8)
```

```python
for wl, psf in zip(wls, psfs):
    y, x = np.indices(psf['OVERDIST'].data.shape)
    dx = x - psf['OVERDIST'].data.shape[1]/2 + 0.5
    dy = y - psf['OVERDIST'].data.shape[0]/2 + 0.5
    r = np.sqrt(dx**2 + dy**2)

    sorti = np.argsort(r.ravel())
    rsort = r.ravel()[sorti]
    psfsorti = psf['OVERDIST'].data.ravel()[sorti]
    ee = np.cumsum(psfsorti)
    plt.plot(rsort, ee/ee[-1], label=f'{wl:.2f}')

    plt.legend(loc=0)

    plt.ylabel('EE')
    plt.xlabel('Radius (pixels)')
    plt.xlim(0,8)
```

Pretty minor variation within wavelength so a coarse grid is probably fine.
