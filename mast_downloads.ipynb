```python
import os
from pathlib import Path

import numpy as np

from astroquery.mast import MastMissions

if 'MAST_TOKEN' in os.environ:
    MastMissions.login(os.environ['MAST_TOKEN'])
else:
    masttokenfile = Path.home() / '.mast_token'
    if masttokenfile.is_file():
        with open(masttokenfile) as f:
            MastMissions.login(f.read().strip())
    else:
        print('No MAST token found. Please set the MAST_TOKEN environment variable or create a .mast_token file in your home directory.')
```

```python
jwst = MastMissions(mission='jwst')
allcam = jwst.query_criteria(program=2609, instrume='nircam')
allspec = jwst.query_criteria(program=2609, instrume='nirspec')
len(allcam), len(allspec)
```

```python
set(allspec['targprop'])
```

```python
def download_data(targetname, dldirname, prodtype='cal'):
    specs = allspec[((allspec['targprop'] == targetname) &
                 (np.strings.count(allspec['productLevel'].filled(''), '3') == 0) & 
                 (allspec['exp_type']=='NRS_MSASPEC'))]
    print(len(specs), 'specs found for target', targetname)

    prods = jwst.get_unique_product_list(specs)
    cals = prods[(prods['file_suffix'] == f'_{prodtype}') & (prods['type']=='science')]

    allsize = np.sum(cals['size'])
    print(f'{len(cals)} {prodtype}s to download, total size {allsize/1e6:.2f} MB')

    return jwst.download_products(cals, download_dir=dldirname, flat=True)
```

```python
download_data('NGC6791-NEWCATALOG-GAIAEDR3', 'ngc6791_cals')
```

```python
download_data('M71-NEWCATALOG-GAIAEDR3', 'm71_cals')
```
